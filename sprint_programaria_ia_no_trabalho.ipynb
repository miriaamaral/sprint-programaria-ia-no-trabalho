```python
# Importamos a linha de montagem e as novas ferramentas de "Auto" carregamento
from transformers import pipeline, AutoTokenizer, AutoModelForQuestionAnswering
```

```python
# 2. Escolhendo qual modelo vamos usar
model_name = "pierreguillou/bert-base-cased-squad-v1.1-portuguese"
```

```python
# 3. Assumindo o controle: Carregamos o Fatiador (Tokenizer) e o Cérebro (Model) manualmente!
meu_fatiador = AutoTokenizer.from_pretrained(model_name)
meu_cerebro = AutoModelForQuestionAnswering.from_pretrained(model_name)
```

```python
# 4. Montando a linha de montagem (pipeline) passando as nossas próprias peças
qa_pipeline = pipeline(
    "question-answering", # Adicionei a tarefa explicitamente aqui
    model=meu_cerebro,
    tokenizer=meu_fatiador
)
```

```python
# 5. O nosso contexto (pode mudar esse texto para testar depois!)
contexto = """
O Brasil é um país da América do Sul.
Capital: Brasília
Moeda: Real
Receita de bolo simples: 2 xícaras de farinha, 3 ovos, 1 xícara de leite
"""
```

```python
# 6. A nossa receitinha para fazer as perguntas
def responder(pergunta):
    try:
        resultado = qa_pipeline(question=pergunta, context=contexto)
        return resultado['answer']
    except Exception as e:
        return f"Erro ao responder: {e}"
```

```python
# 7. Fazendo os testes
print("1:", responder("Qual é a capital do Brasil?"))
print("2:", responder("Quantos ovos vai no bolo?"))
print("3:", responder("Qual é a moeda do Brasil?"))
print("4:", responder("De qual continente o Brasil faz parte?"))
```
